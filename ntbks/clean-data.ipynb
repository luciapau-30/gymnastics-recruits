```python
import pandas as pd
```

```python
fives = pd.read_csv('../data/2025s.csv')
df = fives.copy()
```

```python
df[['Club Gym', 'State']] = df['Gym'].str.extract(r'^(.*?)\s*\((\w+)\)$')
df['Club Gym'] = df['Club Gym'].fillna(df['Gym'])
```

```python
df['Region'] = df['Region'].apply(lambda x: x[1:] if isinstance(x,str) and x.startswith('R') else x)
```

```python
df[['Level', 'Year']] = df['Level'].str.extract(r'^([A-Za-z0-9]+)-(\d{4})$')
```

```python
df_all['Country'] = df_all['State'].apply(lambda x: 'USA' if pd.notna(x) else 'International')

```

```python
df_all['Level'] = df_all['Level'].str.strip().str.upper()
```

```python
df_all['Level'] = df_all['Level'].replace({'E': 'Elite'})
```

```python
print(df_all['Level'].unique())
```

```python
df_all['IsElite'] = df_all['Level'] == 'Elite'
```

```python
df_all['IsElite'] = df_all['Level'] == 'Elite'
```

```python
score_cols = ['VT High', 'UB High', 'BB High', 'FX High', 'AA High']
df_all[score_cols] = df_all[score_cols].apply(pd.to_numeric, errors='coerce')
```

```python
df_all.head()
```

```python
## start rating gymnasts 
```

```python
df_all['Rating'] = (
    0.25 * df_all['VT High'] +
    0.25 * df_all['UB High'] +
    0.25 * df_all['BB High'] +
    0.25 * df_all['FX High']
)
```

```python
df_all['HasPerfect10'] = df_all[['VT High', 'UB High', 'BB High', 'FX High']].eq(10.0).any(axis=1)
df_all['Rating'] += df_all['HasPerfect10'].astype(int)  # +1 if True
```

```python
df_all.sort_values(by='Rating', ascending=False).head(10)
```
